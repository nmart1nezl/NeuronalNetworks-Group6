```python
#Funciones
def df_word(word):
  lines = [p.text.strip() for p in word.paragraphs if p.text.strip() != ""]
  rows = [re.split(r'\t+', line) for line in lines]
#index 1 contain the headers, data starts from index 2
  df = pd.DataFrame(rows[2:], columns=rows[1])
  return df

def normalize_sample_id(sample):
  #isaplha() checks for alphabetic letters
  #this code is very specific for the name of our samples
    if len(sample) == 8 and sample[-1].isalpha():
        return sample[:4] + '-' + sample[4:-1] + '-' + sample[-1]
    else:
        return sample[:4] + '-' + sample[4:]
```

Calcular:
- Alpha diversity (within-sample diversity - observed richness, shannon index, simpson index). <- Does hp reduce diversity?
- Beta diversity (microbial communities cluster by infection status? - bray-curtis distance, jaccard distance, permanova). <- Does hp shift global composition?
- Taxonomic composition differences (mean relative abundance by group, Wilcoxon test per taxon, FDR correction). <- which taxa drive differences?
- Compositional data analysis (CLR transformation, aitchison distance, pca on CLR).
- Relationships with the histological score (Spearman correlation between richness and score, kruskal-wallis test across score groups). <- does inflammation severity track microbiome change?
